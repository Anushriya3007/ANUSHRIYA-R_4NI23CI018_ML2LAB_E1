```python
#read a dataset from the user.(1)use the find-S algorithm to find the most specific hypothesis consistent (2)state the final hypothesis after processing all positive examples
```

```python
import pandas as pd
data = [
    ['Sunny', 'Hot', 'High', 'Weak', 'Yes'],
    ['Sunny', 'Hot', 'High', 'Strong', 'No'],
    ['Rain', 'Cool', 'Normal', 'Strong', 'No'],
    ['Overcast', 'Cool', 'Normal', 'Strong', 'No'],
    ['Sunny', 'Mild', 'Normal', 'Weak', 'Yes'],
    ['Sunny', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Overcast', 'Hot', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Mild', 'High', 'Strong', 'No'],
    ]
```

```python
columns = ['Outlook', 'Temperature', 'Humidity', 'Wind', 'EnjoySport']
df=pd.DataFrame(data , columns=columns)
x=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
hypothesis = None
for i in range(len(x)):
    if y[i] == 'Yes':
      if hypothesis is None:
        hypothesis = x[i].copy()
      else:
        for j in range(len(hypothesis)):
          if hypothesis[j] != x[i][j]:
            hypothesis[j] = '?'
      print("Step", i+1,"Hypothesis:",hypothesis)
print("\nFinal Hypothesis:", hypothesis)
```
