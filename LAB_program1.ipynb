```python
print(df)
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
missing = df.isnull().sum()

plt.figure(figsize=(8, 5))
missing[missing > 0].plot(kind="bar", color="orange")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.show()
```

```python
df["Study_Hours"] = df["Study_Hours"].fillna(
    df["Study_Hours"].median()
)

df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].median()
)

df["Assignments"] = df["Assignments"].fillna(
    df["Assignments"].median()
)
```

```python
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
print("Average Exam Score:", np.mean(df["Exam_Score"]))
print("Median Exam Score:", np.median(df["Exam_Score"]))
print("Maximum Exam Score:", np.max(df["Exam_Score"]))
print("Minimum Exam Score:", np.min(df["Exam_Score"]))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Exam_Score"],
    bins=8,
    edgecolor="black",
    color="skyblue"
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Study_Hours"],
    bins=8,
    edgecolor="black",
    color="lightgreen"
)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Exam_Score"])

plt.title("Box Plot of Exam Scores")
plt.ylabel("Exam Score")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

print(correlation)
```

```python
plt.figure(figsize=(9, 7))

plt.imshow(
    correlation,
    cmap="coolwarm",
    interpolation="nearest"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix")

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Exam_Score"],
    color="purple"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.grid(alpha=0.3)
plt.show()
```
